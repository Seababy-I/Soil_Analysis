```python
from collections import Counter

image_extensions = (".jpg", ".jpeg", ".png")

format_counts = Counter()

for soil_class in soil_classes:
    class_path = os.path.join(IMAGE_DATA_PATH, soil_class)

    for file in os.listdir(class_path):
        if file.lower().endswith(image_extensions):
            extension = os.path.splitext(file)[1].lower()
            format_counts[extension] += 1

print("Image formats found:\n")

for extension, count in format_counts.items():
    print(f"{extension}: {count}")
```

```python
from PIL import Image
from collections import Counter

image_sizes = Counter()

for soil_class in soil_classes:
    class_path = os.path.join(IMAGE_DATA_PATH, soil_class)

    for file in os.listdir(class_path):
        if file.lower().endswith(image_extensions):

            file_path = os.path.join(class_path, file)

            try:
                with Image.open(file_path) as img:
                    image_sizes[img.size] += 1

            except Exception as e:
                print(f"Could not read: {file_path}")

print("Image dimensions found:\n")

for size, count in image_sizes.most_common():
    print(f"{size}: {count} images")
```

```python
corrupted_images = []

for soil_class in soil_classes:
    class_path = os.path.join(IMAGE_DATA_PATH, soil_class)

    for file in os.listdir(class_path):
        if file.lower().endswith(image_extensions):

            file_path = os.path.join(class_path, file)

            try:
                with Image.open(file_path) as img:
                    img.verify()

            except Exception:
                corrupted_images.append(file_path)

print("Number of corrupted images:", len(corrupted_images))

if corrupted_images:
    print("\nCorrupted image files:")
    for image in corrupted_images[:10]:
        print(image)
else:
    print("No corrupted images found.")
```

# Structured Dataset Inspection - WoSIS

This section inspects the WoSIS December 2023 Snapshot dataset.
The dataset consists of multiple tab-separated files containing soil profile information and measured soil properties.

```python
tsv_files = sorted([
    file for file in os.listdir(WOSIS_PATH)
    if file.lower().endswith(".tsv")
])

print("Total TSV files currently available:", len(tsv_files))
print("\nWoSIS TSV files:\n")

for file in tsv_files:
    print(file)
```

```python
keywords = [
    "nit",     # Nitrogen
    "ph",      # pH / phosphorus-related names
    "clay",
    "sand",
    "silt",
    "org",     # Organic properties
    "oc",      # Organic carbon
    "cec",     # Cation exchange capacity
    "bd",      # Bulk density
    "layers"
]

relevant_files = []

for file in tsv_files:
    if any(keyword in file.lower() for keyword in keywords):
        relevant_files.append(file)

print("Potentially relevant files for soil analysis:\n")

for file in relevant_files:
    print("-", file)
```

```python
# Important WoSIS files for initial soil analysis

wosis_selected_files = {
    "layers": "wosis_202312_layers.tsv",
    "clay": "wosis_202312_clay.tsv",
    "sand": "wosis_202312_sand.tsv",
    "silt": "wosis_202312_silt.tsv",
    "organic_carbon": "wosis_202312_orgc.tsv",
    "nitrogen": "wosis_202312_nitkjd.tsv",
    "ph": "wosis_202312_phaq.tsv"
}

for name, filename in wosis_selected_files.items():
    filepath = os.path.join(WOSIS_PATH, filename)
    print(f"{name}: {os.path.exists(filepath)}")
```

```python
# Check whether layer_id is available in all selected tables

for name, filename in wosis_selected_files.items():

    filepath = os.path.join(WOSIS_PATH, filename)

    df_sample = pd.read_csv(
        filepath,
        sep="\t",
        nrows=1000,
        low_memory=False
    )

    print(f"\n{name.upper()}")

    if "layer_id" in df_sample.columns:
        print("✓ layer_id is present")
    else:
        print("✗ layer_id is NOT present")

    if "profile_id" in df_sample.columns:
        print("✓ profile_id is present")
    else:
        print("✗ profile_id is NOT present")
```

```python
# Check duplicate layer_id values in each soil property table

for name, filename in wosis_selected_files.items():

    if name == "layers":
        continue

    filepath = os.path.join(WOSIS_PATH, filename)

    df_temp = pd.read_csv(
        filepath,
        sep="\t",
        usecols=["layer_id", "value_avg"],
        low_memory=False
    )

    total_rows = len(df_temp)
    unique_layers = df_temp["layer_id"].nunique()
    duplicate_rows = total_rows - unique_layers

    print("\n" + "=" * 50)
    print(name.upper())
    print("=" * 50)

    print("Total rows:", f"{total_rows:,}")
    print("Unique layer IDs:", f"{unique_layers:,}")
    print("Additional duplicate measurements:", f"{duplicate_rows:,}")
```

```python
# Function to load and clean a WoSIS soil property table

def load_soil_property(filename, property_name):

    filepath = os.path.join(WOSIS_PATH, filename)

    df = pd.read_csv(
        filepath,
        sep="\t",
        usecols=["layer_id", "value_avg"],
        low_memory=False
    )

    # Rename value_avg to the actual soil property name
    df = df.rename(
        columns={"value_avg": property_name}
    )

    # Remove missing measurement values
    df = df.dropna(subset=[property_name])

    return df
```

```python
clay_df = load_soil_property(
    "wosis_202312_clay.tsv",
    "clay"
)

sand_df = load_soil_property(
    "wosis_202312_sand.tsv",
    "sand"
)

silt_df = load_soil_property(
    "wosis_202312_silt.tsv",
    "silt"
)

organic_carbon_df = load_soil_property(
    "wosis_202312_orgc.tsv",
    "organic_carbon"
)

nitrogen_df = load_soil_property(
    "wosis_202312_nitkjd.tsv",
    "nitrogen"
)

ph_df = load_soil_property(
    "wosis_202312_phaq.tsv",
    "ph"
)

print("All property tables loaded successfully.")
```

```python
property_datasets = {
    "Clay": clay_df,
    "Sand": sand_df,
    "Silt": silt_df,
    "Organic Carbon": organic_carbon_df,
    "Nitrogen": nitrogen_df,
    "pH": ph_df
}

for name, df in property_datasets.items():

    print("\n" + "=" * 50)
    print(name.upper())
    print("=" * 50)

    print("Shape:", df.shape)
    display(df.head())
```

```python
# Load essential soil layer information

layers_path = os.path.join(
    WOSIS_PATH,
    "wosis_202312_layers.tsv"
)

layers_df = pd.read_csv(
    layers_path,
    sep="\t",
    usecols=[
        "profile_id",
        "layer_id",
        "profile_code",
        "layer_name",
        "upper_depth",
        "lower_depth"
    ],
    low_memory=False
)

print("Layers dataset shape:", layers_df.shape)

display(layers_df.head())
```

```python
# Start with the layers dataset

soil_df = layers_df.copy()

# Merge each soil property using layer_id

soil_df = soil_df.merge(
    clay_df,
    on="layer_id",
    how="left"
)

soil_df = soil_df.merge(
    sand_df,
    on="layer_id",
    how="left"
)

soil_df = soil_df.merge(
    silt_df,
    on="layer_id",
    how="left"
)

soil_df = soil_df.merge(
    organic_carbon_df,
    on="layer_id",
    how="left"
)

soil_df = soil_df.merge(
    nitrogen_df,
    on="layer_id",
    how="left"
)

soil_df = soil_df.merge(
    ph_df,
    on="layer_id",
    how="left"
)

print("Combined dataset shape:", soil_df.shape)

display(soil_df.head())
```

```python
# Check missing values in the combined dataset

missing_values = soil_df.isnull().sum()

print("Missing values:\n")

print(
    missing_values[
        [
            "clay",
            "sand",
            "silt",
            "organic_carbon",
            "nitrogen",
            "ph"
        ]
    ]
)
```

```python
# Select records with complete measurements

complete_soil_df = soil_df.dropna(
    subset=[
        "clay",
        "sand",
        "silt",
        "organic_carbon",
        "nitrogen",
        "ph"
    ]
).copy()

print("Original combined dataset:", f"{len(soil_df):,}")
print("Complete records:", f"{len(complete_soil_df):,}")

print(
    "\nPercentage of complete records:",
    round(
        (len(complete_soil_df) / len(soil_df)) * 100,
        2
    ),
    "%"
)
```

```python
# Summary statistics for the complete soil dataset

numeric_columns = [
    "upper_depth",
    "lower_depth",
    "clay",
    "sand",
    "silt",
    "organic_carbon",
    "nitrogen",
    "ph"
]

display(
    complete_soil_df[numeric_columns].describe()
)
```

```python
# Check soil texture consistency

complete_soil_df["texture_total"] = (
    complete_soil_df["clay"]
    + complete_soil_df["sand"]
    + complete_soil_df["silt"]
)

display(
    complete_soil_df[
        ["clay", "sand", "silt", "texture_total"]
    ].head()
)

print(
    complete_soil_df["texture_total"].describe()
)
```

```python
# Check records where soil texture components
# do not approximately sum to 100

valid_texture_df = complete_soil_df[
    complete_soil_df["texture_total"].between(95, 105)
].copy()

print(
    "Complete records:",
    f"{len(complete_soil_df):,}"
)

print(
    "Records with valid texture totals (95–105):",
    f"{len(valid_texture_df):,}"
)

print(
    "Percentage retained:",
    round(
        len(valid_texture_df)
        / len(complete_soil_df)
        * 100,
        2
    ),
    "%"
)
```

```python
# Check minimum and maximum values

columns_to_check = [
    "clay",
    "sand",
    "silt",
    "organic_carbon",
    "nitrogen",
    "ph"
]

range_summary = pd.DataFrame({
    "Minimum": valid_texture_df[columns_to_check].min(),
    "Maximum": valid_texture_df[columns_to_check].max(),
    "Mean": valid_texture_df[columns_to_check].mean(),
    "Median": valid_texture_df[columns_to_check].median()
})

display(range_summary)
```
